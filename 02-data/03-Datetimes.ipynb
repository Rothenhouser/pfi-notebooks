# Zeitstempel und Zeitzonen

In der Standard Library:

- `datetime` für Zeiten, Daten und Zeitspannen
- `zoneinfo` für Zeitzonen
- (`time` nur für Zeiten - kann wenig was datetime nicht kann)

## datetime.datetime
Vollständiger Zeitstempel mit Datum und Uhrzeit, optional mit Zeitzone.

```python
from datetime import datetime, UTC

# aktuelle Zeit mit Zeitzone
now_utc = datetime.now(UTC)
print("Jetzt (UTC):", now_utc)

# feste Zeit ohne Zeitzone
dt = datetime(2025, 1, 14, 15, 30, 0)
print("Feste Zeit (lokal):", dt)
```

### Umwandeln von/nach String

```python
# ISO 8601 hat feste Methoden:
print("ISO Format:", dt.isoformat())

dt2 = datetime.fromisoformat('2025-12-01 12:00Z')
print("Parsed von ISO Format:", dt2)
```

```python
# Beliebige Formate über Format Codes
print("Datum + Zeit:", dt.strftime("%d.%m.%Y %H:%M:%S"))

dt_parsed = datetime.strptime("14.01.2025 15:30", "%d.%m.%Y %H:%M")
print("Parsed dt:", dt_parsed)
```

Alle Format Codes here: https://docs.python.org/3/library/datetime.html#strftime-and-strptime-format-codes

```python
# Oder über die Attribute von dt:
print(f"Heute ist der {now_utc.day}.{now_utc.month}., nach {now_utc.hour} Uhr.")
```

## datetime.date
Nur das Datum, keine Uhrzeit.

```python
from datetime import date

today = date.today()
print(today)

d = date(2025, 1, 14)
print(d)

# Ist auch eine Methode von datetime Objekten
print(dt.date())
```

## datetime.time
Nur die Uhrzeit, ohne Datum.

```python
from datetime import time

t = time(15, 30, 0)
print("Feste Uhrzeit:", t)

# Mit Mikrosekunden
t2 = time(15, 30, 0, 500000)
print("Mit Mikrosekunden:", t2)
```

## datetime.timedelta
Zeitspannen können direkt erzeugt werden oder ergeben sich aus der Subtraktion von Zeitstempeln:

```python
from datetime import timedelta

td = now_utc - dt2   # Beide müssen Zeitzoneninfo haben oder nicht haben!
print(td)
```

```python
delta = timedelta(days=1, hours=2)

# Addition/Subtraktion
future = now_utc + delta
print(future)

past = now_utc - delta
print(past)

print(f"delta in Stunden: {delta.total_seconds() / 3600}")
```

## Zeitzonen

Beliebige Zeitzonen sind in `zoneinfo` abgebildet.

Dafür muss eventuell das Paket `tzdata` installiert werden, wenn das System keine Zeitzonendatenbank enthält.

```python
from zoneinfo import ZoneInfo

now_local = datetime.now(tz=ZoneInfo("Europe/Berlin"))
print(now_local)

# Die UTC-datetime von oben bezeichnet einen ähnlichen Zeitpunkt:
print("kleiner Abstand:", now_local - now_utc)
```

Zeitzonen umwandeln ist tückisch! Zwei Fälle müssen immer unterschieden werden:
1. Die Zeitzone soll **ersetzt** werden, aber die Uhrzeit ("an der Wanduhr") bleibt unverändert.
2. Die Uhrzeit wird in eine andere Zeitzone **umgewandelt** (die Wanduhr muss umgestellt werden).

```python
# Beispiel: Wir wissen, dass ein System Daten in lokaler Zeitzone schreibt,
# und wollen das richtig einfangen (Fall 1).
dt_str = "2025-12-01 14:30"
dt_naive = datetime.fromisoformat(dt_str)
dt_with_tz = dt_naive.replace(tzinfo=ZoneInfo("Europe/Berlin"))
print("Zeitstempel korrekt eingelesen:", dt_with_tz)
```

```python
# Dann wollen wir die Daten als UTC weiterverarbeiten (Fall 2).
dt_utc = dt_with_tz.astimezone(ZoneInfo("UTC"))
dt_utc
```
