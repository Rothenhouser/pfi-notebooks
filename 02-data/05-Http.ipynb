# Daten von HTTP APIs

`requests` ist der Standard für Requests an HTTP APIs. Die üblichsten Operation sind GET (um Daten herunter zu laden) und POST (um Daten zu senden).

```python
import requests

# Schätze das Alter meines Namens.

# Doku unter https://agify.io/documentation.
url = 'https://api.agify.io'
response = requests.get(url, params={'name': 'Nicolas', 'country_id': 'DE'})
print("Status Code:", response.status_code)
print("Raw Content:", response.text)
```

```python
# Response ist JSON und kann geparsed werden:
data = response.json()

print("Name:", data['name'])
print("Geschätztes Alter:", data['age'])
```

## Status Codes prüfen
Ein Fehlercode vom Server erzeugt (erstmal) keinen Fehler in Requests.

Meistens nützlich, das direkt zu prüfen:

```python
def get_with_error(url):
    response = requests.get(url)
    response.raise_for_status()
    # nur wenn erfolgreich
    return response

get_with_error('https://api.agify.io?name=Nicolas&country_id=DE')
# get_with_error('https://api.agify.io/nonexistent')
```
