# Mehr Pandas Input/Output

## CSV

CSV ist, anders als JSON,  kein standardisiertes Format. Pandas hat dafür eine sehr flexible `pd.read_csv` Funktion.

Fast alle tabellarischen Textdaten können irgendwie eingelesen werden.

```python
import pandas as pd
from pathlib import Path

# Ein schwieriges CSV-File.
german_csv = Path('./data/taxi_de.csv')

pd.read_csv(german_csv)
```

Der Default funktioniert hier nicht. Wir haben:
- eine Kommentarzeile, dann eine leere Zeile
- noch eine Kommentarzeile unter dem Header
- Semikolon als Trenner
- Dezimalkomma
- ganz fehlende Werte
- -999 als Indikator für fehlende Werte

In einem automatisierten Prozess kann das File nicht vorher von Hand bereiningt werden.

```python
# Besser:
taxi_df = pd.read_csv(
    german_csv,
    skiprows=[0, 1, 3],
    sep=';',
    decimal=',',
    na_values=[-999],
    index_col='Datum',
    parse_dates=True,
    date_format='%d.%m.%Y'
)
taxi_df.head()
```

Dokumentation ist hier: https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html

## Excel-Files lesen und schreiben

Pandas kann Excel-Files schreiben, und einfach genug strukturierte auch einlesen.

Pandas greift dabei auf das Paket `openpyxl` zurück.

```python
taxi_df.to_excel('taxi_example.xlsx', sheet_name='made up data')
```

```python
pd.read_excel('taxi_example.xlsx').head()
```
