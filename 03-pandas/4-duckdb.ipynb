# DuckDB

DuckDB (https://duckdb.org/docs/) ist eine in-process SQL-Datenbank für analytische Aufgaben.

Usecases:
1. Man schreibt lieber SQL als pandas.
2. Die Daten passen nicht mehr in RAM.

Das Pythonpaket `duckdb` enthält die komplette Engine.

```python
import duckdb

con = duckdb.connect(':memory:')

query = """
SELECT
    DATE_TRUNC('month', tpep_dropoff_datetime) AS month,
    AVG(total_amount) AS avg_fare,
    COUNT(*) AS num_trips
FROM read_csv_auto('./data/yellow_tripdata_2025.csv')
GROUP BY month
ORDER BY month;
"""
monthlies = con.execute(query).df()
monthlies
```

DuckDB kann auch Queries auf Dataframes in Memory ausführen. Die Daten werden dabei nicht kopiert.

```python
import pandas as pd

meteorites = pd.read_csv('./data/Meteorite_Landings.csv')

# Direkter Zugriff auf den in-memory dataframe:
duckdb.sql(
    """
    SELECT * FROM meteorites
    WHERE "mass (g)" > 100_000
    """
)
```
