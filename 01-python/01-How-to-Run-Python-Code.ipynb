Warum wird es so kompliziert?

Es kann viele verschiedene `python.exe` geben:
- eine oder mehrere System-Versionen, diese sind z.B. von python.org oder über dem Microsoft Store installiert
- Kopien davon in verschiedenen, projektspezifischen *Virtual Environments* 

Externe Pakete können über verschiedene Paketmanager installiert werden: pip, conda, uv, ...


![Virtual envs on Windows](fig/venvs-windows.svg)

Welches Python als Default ausgeführt wird, hängt von der Umgebungsvariable PATH ab.

In Powershell:
```powershell
> where.exe python
```

```powershell
> $env:Path
```

Jedes python.exe kann aber auch direkt mit dem vollen Pfad ausgeführt werden.


Warum Virtual Environments?
- keine Konflikte zwischen Projekten 
- kein "zumüllen" des System-Python
- Reproduzierbarkeit auf anderen Rechnern (wenn Pakete z.B. in einem requirements.txt definiert werden)

Es gibt,  historisch gewachsen, viele mögliche Pythoninstallationen und Paketmanager für Python.

Ein Standardsetup wäre etwa:

1. Wenn noch kein Python: Stabile Pythonversion von python.org installieren.
2. Terminal im Projektordner starten und sicherstellen, dass das neue Python im PATH ist: `where.exe pip`.
2. Virtual Environment anlegen: `python -m venv .venv`
4. Virtual Environment aktivieren:`.venv\Scripts\activate.bat` wenn in CMD, `.venv\Scripts\activate.ps1` wenn in Powershell
4. Prüfen, dass jetzt das Python aus dem  Virtual Environment im PATH ganz oben ist: `where.exe pip`
5. Zusätzliche Pakete installieren: `python -m pip install jupyter pandas`...

## Ausführen - Python Interpreter

Aus Benutzersicht muss Python nicht kompiliert werden, sondern ist eine interpretierte Sprache.

Mit dem Interpreter können wir in der Kommandozeile interaktiv arbeiten.

Dazu führen wir - mit dem Virtual Environment immer noch aktiviert - auf der Kommandozeile `python` aus:



```cmd
>python
Python 3.12.3 (tags/v3.12.3:f6650f9, Apr  9 2024, 14:05:25) [MSC v.1938 64 bit (AMD64)] on win32
Type "help", "copyright", "credits" or "license" for more information.
>>>
```


Im Interpreter lassen sich sehr praktisch kleinere Operationen ausprobieren.

Beispiel als Taschenrechner:
``` python
>>> 1 + 1
2
>>> x = 5
>>> x * 3
15
```

#### Übung

- Gelten die normalen Punkt vor Strich Regeln?
- Wieviel ist 2 hoch 10?

<details>

<summary> Tipps </summary>

- Potenz-Operator ist `**`.
</details>

## Skripte

Code wird in einem Textfile mit der Endung `.py` gespeichert:

``` python
# file: test.py
print("Running test.py")
x = 5
print("Result is", 3 * x)
```

File im Terminal ausführen:
```powershell
>python test.py
Running test.py
Result is 15
```

## Jupyter Notebooks

Ein Zwischending ist das *Jupyter Notebook*. Es ist ein Dokumentenformat, in dem sich ausführbarer Code, formatierter Text, Grafiken, und sogar interaktive Features kombinieren lassen.

Ein Notebook kann Schritt für Schritt Ergebnisse anzeigen, aber auch wie ein Skript gespeichert und als ganzes ausgeführt werden.

Es eignet sich u.a. um Daten zu analysieren, Experimente zu beschreiben oder Skripte zu entwerfen.

Installation im Projekt-Virtual Environment mit:

`python -m pip install jupyter`

Ausführen des Notebook-Servers mit VSCode oder:

`jupyter notebook`

Wichtigste Kommandos (siehe auch "Help" → "Keyboard Shortcuts"):

- `Enter`: Zelle bearbeiten (\"Edit mode\", Zellernrahmen wird grün)
- `Esc`: Befehlsmodus (\"Command mode\", Zellenrahmen wird blau)
- `Strg+Enter`: Zelle ausführen
- `Shift+Enter`: Zelle ausführen und nächste aktivieren
- `A`: Neue Zelle über (**a**bove) der aktuellen einfügen
- `B`: Neue Zelle unter (**b**elow) der aktuellen einfügen
- `M`: Zellentyp = Markdown (Text)
- `Y`: Zellentyp = Code
- `H`: Kurzbefehle anzeigen
